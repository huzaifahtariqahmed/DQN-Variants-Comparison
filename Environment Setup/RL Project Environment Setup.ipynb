### **Import Libraries**

```python
%matplotlib inline
import gym
import gymnasium as gym
import numpy as np
from gym.wrappers.record_video import RecordVideo
import glob
import io
import base64
from IPython.display import HTML
from pyvirtualdisplay import Display
from IPython import display as ipythondisplay
from matplotlib import pyplot as plt
from IPython.display import HTML, display
import time
from scipy.integrate import quad
from collections import defaultdict
import pygame
import os
import time
from moviepy.editor import ImageSequenceClip
import base64
```

### **Environment Creation**

```python
env = gym.make("CartPole-v1", render_mode="rgb_array")
```

In this step, we create the CartPole-v1 environment and start collecting frames for video recording. We specify the render mode as "rgb_array", which allows us to capture each frame as an image.

**Rationale:**  
- `CartPole-v1` has discrete actions and runs quickly.  
- `render_mode='rgb_array'` is required to capture frames via code.

**Observation:**  
Environment reset and initial render returned valid state vectors and image arrays.

### **Running Environment on Random Actions**

```python
frames = []
obs, info = env.reset()
n_frames = 150    # 150 frames @ 30 fps ⇒ 5 s
for i in range(n_frames):
    frame = env.render()      # returns an RGB array
    frames.append(frame)
    action = env.action_space.sample()
    obs, reward, terminated, truncated, info = env.step(action)
    if terminated or truncated:
        obs, info = env.reset()

env.close()
```

We run the environment for 150 steps (which should last about 5 seconds at 30 frames per second). The agent takes random actions in each step. For each action, we capture the frame and append it to the frames list.

**Challenge:**

Rendering Speed: At first, the video was being captured too quickly, and the video duration was too short. To address this, we ensured a sufficient number of steps and slowed down the process if necessary.

### **Displaying Recorded Video**

Once we have collected enough frames, we use MoviePy to combine these frames into a video file. The video is encoded and saved as an MP4 file with a frame rate of 30 FPS. The result is a 5-second video of the agent's actions.

```python
os.makedirs("video_manual", exist_ok=True)
clip = ImageSequenceClip(frames, fps=30)
video_path = "video_manual/cartpole_random.mp4"
clip.write_videofile(video_path, codec="libx264", verbose=False, logger=None)
```

Finally, we read the video file and embed it in the notebook using HTML so we can view it directly in the notebook interface.

```python
video_data = open(video_path, "rb").read()
b64 = base64.b64encode(video_data).decode("ascii")
HTML(f"""
<video width=480 controls>
  <source src="data:video/mp4;base64,{b64}" type="video/mp4">
</video>
""")
```

### **Challenges & Observations**

**Challenges:**

- Short Video Duration: The video was initially too short due to the agent quickly reaching the termination state in CartPole. We had to enforce a minimum number of frames (steps) to ensure a longer video duration.

- Rendering Speed: Even after adjusting the number of steps, the frame rate of the environment rendered the video too quickly. By introducing a small delay between each action (time.sleep()), we slowed the agent's actions to allow for smoother frame capture.

**Observations:**

- The CartPole environment is fairly easy for a random agent, with episodes typically ending within a few dozen steps.

- Capturing a video of the agent's actions provides a useful way to visualize the agent-environment interaction and debug the agent's behavior.
